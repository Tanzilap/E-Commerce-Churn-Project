## RFM Analysis

##### R -- Recency(Days Since the last purchase)
##### F -- Frequency(Total number of purchases)
##### M -- Monetary(Total amount spent)

```python
# First, calculate the max invoice date once
max_date = df['invoice_date'].max()

# Then group and aggregate
rfm = df.groupby('customer_id').agg(
    Frequency=('invoice_date', 'count'),
    Monetary=('price', 'sum'),
    LastPurchaseDate=('last_purchase_date', 'max')
).reset_index()

# Now calculate Recency
rfm['Recency'] = (max_date - rfm['LastPurchaseDate']).dt.days

# Drop LastPurchaseDate if no longer needed
rfm = rfm.drop(columns=['LastPurchaseDate'])

# Display result
print(rfm.head())
```

```python
# Assign scores (higher score = better customer)
rfm['R_Score'] = pd.qcut(rfm['Recency'], 5, labels=[5, 4, 3, 2, 1])
rfm['F_Score'] = pd.qcut(rfm['Frequency'], 5, labels=[1, 2, 3, 4, 5])
rfm['M_Score'] = pd.qcut(rfm['Monetary'], 5, labels=[1, 2, 3, 4, 5])

# Combine RFM scores
rfm['RFM_Score'] = rfm['R_Score'].astype(str) + rfm['F_Score'].astype(str) + rfm['M_Score'].astype(str)

# Display RFM segmentation
print(rfm.head())
```

## K-Clustering

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()
scaled=scaler.fit_transform(rfm)
```

```python
from sklearn.cluster import KMeans

# Initialize list to store inertia values
inertia=[]
for i in np.arange(1,11):
    kmeans=KMeans(n_clusters=i)
    kmeans.fit(scaled)
    inertia.append(kmeans.inertia_)
    
# Plot the elbow curve
plt.plot(inertia,marker='o')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.show()
```

optimal cluster k =3

```python
# Fit KMeans with 3 clusters
kmeans=KMeans(n_clusters=3)
kmeans.fit(scaled)

# Assign cluster labels to RFM, shifting by +1 to start clusters from 1
rfm['Clusters']=(kmeans.labels_ +1)

# Preview the updated RFM table
rfm.head()
```

```python
# Analyze the average RFM values for each cluster
cluster_summary = rfm.groupby('Clusters').agg({'Recency': 'mean', 'Frequency': 'mean', 'Monetary': 'mean'})
print(cluster_summary)
```

```python
# Function to assign customer segment labels based on cluster value
def func(row):
    if row['Clusters']==1:
        return 'Moderate Spenders'
    elif row['Clusters']==2:
        return 'Lapsed Customer'
    else:
        return 'High Spenders'
```

```python
# Apply the segmentation function to create a new 'Conditions' column
rfm['Conditions'] = rfm.apply(func , axis = 1)

# knowing the percentage of each segment of customers using pie chart
segment_count = rfm['Conditions'].value_counts()

# Plotting the pie chart
plt.pie(segment_count.values, labels= segment_count.index, autopct='%1.1f%%')
plt.title('Percentage of Customer in each Segment')
plt.show()
```

### RFM Analysis Conclusion
The customer base has been segmented into three groups using RFM metrics and clustering. The distribution of customers across segments is as follows:

Moderate Spenders: 38.4%
This is the largest segment. These customers show average behavior in terms of spending and engagement. They may be influenced with personalized offers to increase loyalty and spend.

High Spenders: 33.6%
A significant portion of the customer base falls under this category. These are highly valuable customers, likely with recent and frequent purchases. Efforts should focus on retention and VIP treatment (e.g., early access, exclusive deals).

Lapsed Customers: 27.9%
Nearly one-third of customers are at risk or inactive. These customers require re-engagement strategies, such as win-back emails, discount incentives, or targeted campaigns.
